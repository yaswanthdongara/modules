```python
lst=["mamithabaiju",22,6,2001,"kalyani priyadarshan",5,4,1993]
df=pd.DataFrame(lst,index=["fav1","dob-d","dob-m","dob-y","fav2","dob-d","dob-m","dob-y"])
print(df)
print(df.dtypes,type(df))
```

```python
lst=[["fav1","fav2"],[25,12]]
df = pd.DataFrame(lst)
print(df)
```

```python
lst=[["fav1","fav2","fav3"],["kalyani priyadarshan","mamitha baiju","phoebe"]]
df = pd.DataFrame([lst[1]],columns=[lst[0][i] for i in range(len(lst[0]))])
print(df)
```

```python
lst=[["a","b","a+b"],["r1","r2","r3"],[[12,25,37],[17,18,35],[33,55,88]]]
df = pd.DataFrame([lst[2][0],lst[2][1],lst[2][2]],columns=[lst[0][i] for i in range(len(lst[0]))],index=[lst[1][i] for i in range(len(lst[1]))])
print(df)
```

```python
d={"fav":["fav1","fav2","fav3"],"name":["mamithabaiju","phoebe","kalyani priyadarshan"]}
df = pd.DataFrame(d)
print(df)
```

```python
a = np.array([["fav1","fav2","fav3"],["mamithabaiju","phoebe","kalyani priyadarshan"]])
df = pd.DataFrame(a)
print(df)
```

```python
a = np.array([["fav1","fav2","fav3"],["mamithabaiju","phoebe","kalyani priyadarshan"]])
df = pd.DataFrame([val for val in a[1]],index=[val for val in a[0]])
print(df)
```
