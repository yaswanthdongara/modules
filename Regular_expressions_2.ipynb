| Category | Regex | Meaning |
|---|---|---|
| Characters | `.` | Any character |
| Characters | `[abc]` | `a`, `b`, or `c` |
| Characters | `[a-z]` | Lowercase letters |
| Characters | `[A-Z]` | Uppercase letters |
| Characters | `[0-9]` | Digits |
| Characters | `[^abc]` | Anything except `a`, `b`, `c` |
| Shortcuts | `\d` | Digit |
| Shortcuts | `\D` | Not a digit |
| Shortcuts | `\s` | Whitespace |
| Shortcuts | `\S` | Not whitespace |
| Shortcuts | `\w` | Word character |
| Shortcuts | `\W` | Not a word character |
| Quantifiers | `*` | 0 or more |
| Quantifiers | `+` | 1 or more |
| Quantifiers | `?` | 0 or 1 |
| Quantifiers | `{n}` | Exactly n |
| Quantifiers | `{n,}` | n or more |
| Quantifiers | `{n,m}` | n to m |
| Positions | `^` | Beginning |
| Positions | `$` | End |
| Positions | `\b` | Word boundary |
| Logic | `|` | OR |
| Groups | `()` | Group |
| Lookaround | `(?=...)` | Followed by |
| Lookaround | `(?!...)` | Not followed by |
| Lookaround | `(?<=...)` | Preceded by |
| Lookaround | `(?<!...)` | Not preceded by |

```python
#example 1
#extracting names that starts with capital letter
import re
st = "PhoebeLitchfield is my favorite crickter, MamithaBaiju, KalyaniPriyadarshan are popular in malayalam movies"
s = '[A-Z][a-z]+' #One uppercase letter followed by one or more lowercase letters.
names = re.findall(s,st)
print(names)
```

```python
#example 2
#extracting names from emails
import re
with open('/content/mails.txt','r') as f:
    st = f.read()
print("-"*70)
print("Original File Data:",st)
print("-"*70)
s = r'[^@\s]+(?=@)'
names = re.findall(s,st)
print("extracting names from user emails")
print("-"*70)
print("extracted data")
print(names)
print("-"*70)
```
