```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("students_marks_25_rows.csv")
print(df)
```

```python
#adding a new column by an expression
df["percentage"] = (df["Mathematics"] + df["Science"] + df["English"] + df["Computer Science"] + df["Social Studies"])/5
print(df)
```

```python
#sort the dataframe data on col name percentage
df.sort_values(["percentage"])
```

```python
df[["Student Name","percentage"]].iloc[-1]
```

```python
df.sort_values(["percentage"],ascending=False).head()
```

```python
print(df.head())
```

after sorting the data, the result is not permanent, to make it permanent we use inplace=True

```python
df.sort_values(["percentage"],ascending=False,inplace=True)
```

```python
print(df.head())
```

```python
df["Grade"] = "A+"
print(df.head())
```

```python
#removing a column
df.drop(columns=["Grade"],inplace=True)
print(df.head())
```

to make changes permanently use inplace=True

```python
df.drop(columns=["percentage"])
print(df.head())
```

by default inplace=False

```python
df.drop(columns=["percentage","Mathematics"])
print(df.head())
```

```python
df.drop(columns=df.columns[2]).head()
```

```python
#dropping columns by col_number
df.drop(df.columns[2],axis=1).head()
```

```python
df["cgpa","sgpa"] = None
print(df.head())
```

```python
df.drop(index=10)
```

```python
df.drop(index=10,axis=1,inplace=True)
print(df)
```

```python
df.drop(index=[6,19],axis=0,inplace=True)
print(df)
```

```python
df.drop(index=[6,19],axis=0,inplace=True)
print(df)
```

```python
df.drop(index=[16,11,13,14,15],axis=0,inplace=True)
print(df)
```

```python
df.drop(columns=["Social Studies","Science","Computer Science","percentage"],inplace=True)
print(df)
```

```python
df.drop(('cgpa', 'sgpa'), inplace=True, axis=1)
print(df)
```
