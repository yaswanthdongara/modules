```python
import numpy as np
import pandas as pd
```

```python
s=pd.Series([12,12,12,25,25,25,33,37,18,17,7,45,3,1,None])
print(s)
```

```python
#sum
sum = s.sum()
print(sum)
```

```python
#mean
size = len(s)
mean = sum/size
print(mean)
mean = s.mean()
print(mean)
mean=s.mean(skipna=False)
print(mean)
```

```python
#mode
print(s.mode())
```

```python
#median
a = pd.Series([7,1,12,25,33,37,55])
print(a.sort_values().values)
print(a.median())
b = pd.Series([7,1,12,25,33,37,55,3])
print(b.sort_values().values)
print(b.median())
print((25+12)/2)
c = pd.Series([7,1,12,25,33,37,55,None])
print(c.median())
print(c.median(skipna=False))
```

```python
#standard deviation
a = pd.Series([7,1,12,25,33,37,55,None])
print(a.std())
```

```python
#aggregation - agg()
a = pd.Series([7,1,12,25,25,33,37,55,None])
print(a.agg(["mean","sum","min","max",lambda x: x.sum()/len(x)]))
```

```python
#descriptive statistics
b=pd.Series([7,12,12,25,25,23,25,37,33,55,63])
print(b.describe())
```

```python
b=pd.Series(["pheobe","litchfield","kalyanipriyadarshan","mamitha.baiju"])
print(b.describe())
```

```python
b=pd.Series(["pheobe","litchfield","kalyanipriyadarshan","mamitha.baiju","pheobe","mamitha.baiju"])
print(b.describe(include="all"))
```

```python
b=pd.Series([1,2,3,4,5])
print(b.cumprod())
```
