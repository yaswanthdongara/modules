```python
import pandas as pd
import numpy as np
```

Adding a Single value to Series

```python
s = pd.Series([25,"Srashta"])
print(s.size)
```

```python
s[s.size] = "panda"
```

```python
print(s)
```

```python
s[len(s)] = "kalyanipriyadarshan"
```

```python
for val in s:
  print(val)
```

```python
s = pd.Series(["kalyanipriyadarshan","mamithabaiju"],index=["fav1","fav2"],name="fav actress")
```

```python
print(s)
```

```python
s["fav3"] = "aneswara.rajan"
s["fav4"] = "samantha"
print(len(s))
print(s)
s.drop("fav4")
print(s)
print(s.size)
s["fav5"] = "hedgepooja"
s.drop(["fav4","fav5"],inplace=True) #default inplace = False
print(s)
print(s.size)
```

Addition of series object

```python
s1=pd.Series([1,2,3,4])
s2=pd.Series([5,6,7,8])
s3=s1.add(s2)
print(s3)
```

```python
s1=pd.Series([1,2,3,4],index=["y","b","c","d"])
s2=pd.Series([5,6,7,8],index=["b","y","z","w"])
s3=s1.add(s2)
print(s3)
```

```python
s1=pd.Series([1,2,3,4],index=["y","b","c","d"])
s2=pd.Series([5,6,7,8],index=["b","y","z","w"])
s3=s1.add(s2,fill_value=0)
print(s3)
```

```python
s1=pd.Series([12,25,33,18])
s2=pd.Series([25,12])
s3=s1.add(s2)
print(s3)
s4=s2.add(s1,fill_value=0).astype(int)
print(s4)
```

```python
s1=pd.Series([12,25,33,37,None,pd.NA,np.nan])
print(s1.isna())
print("-"*70)
s2=s1[s1.isna()]
print(s2)
print("-"*70)
print(s1[s1.isnull()])
```
