```python
import numpy as np
```

```python
a = np.random.randint(1,10,size=9).reshape(3,3)
b = np.random.randint(1,10,size=9).reshape(3,3)
c = np.dot(a,b)
print(a)
print(b)
print(c)
d = a@b
print(d)
```

```python
#inner product
a = np.random.randint(1,10,size=(3,3))
b = np.random.randint(1,10,size=(3,3))
c = np.inner(a,b)
d = np.dot(a,b)
print(a)
print(b)
print(c)
print(d)
```

```python
a = np.random.randint(1,10,size=(2,2))
b = np.random.randint(1,10,size=(2,2))
c = np.outer(a,b)
print(a)
print(b)
print(c)
```

```python
#solving linear equations
a = np.array([
    [2,3],
    [4,1]
])
b = np.array([6,7])
x,y = np.linalg.solve(a,b) #2x+3y = 6, 4x+y = 7
print(x,y)
```

```python
a = np.array([
    [2,1,1],
    [1,2,1],
    [1,1,2]
])
b = np.array([10,11,12])
x,y,z = np.linalg.solve(a,b)
print(x,y,z)
```

```python
#trace of a matrix = sum of all main diagonal elemenets
import numpy as np
a = np.random.randint(1,9,size=(2,2))
print(a)
trace = np.trace(a)
print(trace)
```

```python
import numpy as np
a = np.random.randint(2,9,size=(2,2))
print(a)
s = np.sum(np.diag(a))
print(s)
```

```python
import numpy as np
a = np.random.randint(2,9,size=(3,3))
print(a)
det = np.linalg.det(a)
print(det)
```

```python
import numpy as np
a = np.random.randint(2,9,size=(3,3))
print(a)
inv = np.linalg.inv(a).astype(int)
print(inv)
```

```python
a = np.array([1,2,3,0,1,4,5,6,0]).reshape(3,3)
a_inv = np.linalg.inv(a)
print(a)
print(a_inv)
print((a@a_inv).astype(int))
```

```python
#to save more than one nd array object --- using savez()
import numpy as np
a = np.random.randint(1,10,9)
b = np.random.randint(1,10,size=(3,3))
print(a)
print(b)
np.savez(r"C:\Users\yaswa\Downloads\data.npz",a,b)
print("Data Saved")
```

```python
c = np.load(r"C:\Users\yaswa\Downloads\data.npz")
print(c.files)
print(c["arr_0"])
print(c["arr_1"])
```

```python
#saving multiple nd arrays using savez
a = np.random.randint(1,10,9)
b = np.random.randint(1,10,size=(3,3))
c = np.random.randint(1,10,size=(3,3,3))
print(a)
print(b)
print(c)
np.savez(r"data.npz",a=a,b=b,c=c)
print("Data Saved")
```

```python
#loading savez files
npzfile = np.load("data.npz")
print(npzfile.files)
print(npzfile["a"])
print(npzfile["b"])
print(npzfile["c"])
```

```python
#savez_compressed
a = np.random.randint(1,10,9)
b = np.random.randint(1,10,size=(3,3))
np.savez_compressed(r"data.npz",a=a,b=b)
print("Data Saved")
```

```python
#opening savez_compressed files
c = np.load(r"data.npz")
print(c.files)
print(c["a"])
print(c["b"])
```
