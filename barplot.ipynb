```python
import pandas as pd
import matplotlib.pyplot as plt
savings = [20,30,40,50]
employess = [105,193,38,73]
plt.bar(savings,employess,width=3)
plt.show()
```

```python
A = ["a","b","c"]
B = [10,15,20]
plt.bar(A,B,bottom=5)
plt.show()
```

```python
A = ["a","b","c"]
B = [10,15,20]
plt.bar(A,B,bottom=5,align="edge")
plt.show()
```

```python
A = ["a","b","c"]
B = [10,15,20]
plt.bar(A,B,edgecolor="r")
plt.show()
```

```python
A = ["a","b","c"]
B = [10,15,20]
plt.bar(A,B,edgecolor="r",linewidth=10)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
savings = [20,30,40,50]
employess = [105,193,38,73]
plt.barh(savings,employess)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
savings = [20,30,40,50]
employess = [105,193,38,73]
plt.barh(savings,employess,height=5,color="g")
plt.show()
```

```python
brands=["samsung","vivo","realme","redmi","oppo"]
sales=[12000,25000,15000,14000,10000]
colors=["b","g","r","c","m"]
plt.bar(brands,sales,color=colors)
plt.show()
```

```python
brands=["samsung","vivo","realme","redmi","oppo"]
sales=[12000,25000,15000,14000,10000]
colors=["b","g","r","c","m"]
labels=["samsung","vivo","realme","redmi","oppo"]
plt.bar(brands,sales,color=colors,label=labels)
plt.legend()
plt.show()
```
