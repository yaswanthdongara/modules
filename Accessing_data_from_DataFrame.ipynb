Accessing the data from DataFrame

```python
import pandas as pd
import numpy as np
```

```python
csv = pd.read_csv("/content/AI_Tools_People_Ratings_100x5.csv")
df = pd.DataFrame(csv)
print(df.head())
```

```python
print(df.tail())
```

```python
print(df.tail(2))
```

```python
print(df.to_string())
```

```python
pd.set_option('display.max_rows', None)
print(df)
```

```python
print(df.describe())
```

```python
print(df.shape)
print("No.of rows",df.shape[0])
print("No.of columns",df.shape[1])
```

```python
for col in df.columns:
  print(col)
```

```python
#retreiving specified column
lst = list(df['Name'])
print(lst)
```

```python
for i in range(12):
  print(lst[i].upper())
```

```python
print(df[['Name','ChatGPT']].head())
```

```python
slicing = df[1:26:2]
print(slicing)
```

```python
print(df['Name'][25:26])
```

```python
#retrieving specified row values
print(df[[col for col in df.columns]][12:25:7])
```

```python
print(df[['ChatGPT','Gemini']].tail())
```

```python
pd.set_option("display.max_rows",10)
print(df)
```

```python
print(df.loc[12])
```

```python
print(df.iloc[25])
```

```python
print(df.loc[37,['Name','Gemini','Claude']])
```

```python
print(df[1:26:2][['Name','ChatGPT']])
```

```python
print(df.iloc[:97:-1])
```

```python
print(df.loc["Name":"Copilot"])
```

```python
print(df.loc[1:3])
```

```python
print(df.iloc["Name":"Copilot"])
```
