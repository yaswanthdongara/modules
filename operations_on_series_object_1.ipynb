```python
import numpy as np
import pandas as pd
```

```python
s=pd.Series([12,25,np.nan,pd.NA,None])
print("No.of missing values",s[s.isnull()].size)
print("No.of missing values",s.isnull().sum())
print("No.of missing values",s.isna().sum())
print("No.of missing values",s[s.isna()].size)
```

```python
s=pd.Series([12,25,np.nan,pd.NA,None])
print("No.of missing values",s[s.notnull()].size)
print("No.of missing values",s.notnull().sum())
print("No.of missing values",s.notna().sum())
print("No.of missing values",s[s.notna()].size)
print(s[s.notna()])
print("-"*70)
print(s.notna())
```

```python
#sort the series object values
import numpy as np
a=pd.Series([6, 1, 8, 5, 2, 8, 8, 6, 6])
print(a.values) #if .values is not used then index will also be printed
a.sort_values()
print(a.values)
print(a.sort_values().values)
a.sort_values(inplace=True) #index stays same
print(a)
a.sort_values(ascending=False,inplace=True)
print(a)
```

```python
print(a.values[-1]) #smallest value when 'a' is sorted in descending order
print(a.min())
print(a.values[0]) #largest value when 'a' is sorted in descending order
print(a.max())
```

```python
print(len(a))
print(a.unique())
print(a.nunique())
```

```python
#frequency
print(a.values)
print(a.value_counts().values)
print("-"*70)
print(a.value_counts())
```
