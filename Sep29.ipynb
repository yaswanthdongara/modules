```python
import pandas as pd
import numpy as np
```

```python
s = pd.Series(["yaswanth","srashta","immanuel","beulah","prudhvi"])
print(s,type(s))
```

```python
#index
s = pd.Series(["yaswanth","srashta","immanuel","beulah","prudhvi"],index=["Name "+str(i) for i in range(1,6)])
print(s)
```

```python
#ndim
print(s.ndim,"dimension")
```

```python
#dtype
print("type of s =",s.dtype)
```

```python
#size and shape
print("No.of elements",s.size)
print("shape of s",s.shape)
```

```python
#name, index
s = pd.Series(["yaswanth","srashta","immanuel","beulah","prudhvi"],index=["Name "+str(i) for i in range(1,6)],name="Family")
print(s.name)
for index in s.index:
  print(index)
```

```python
#values
print(s.values,type(s.values))
for value in s.values:
  print(value.upper())
```

```python
#has nan values
print(s.hasnans)
s = pd.Series(["yaswanth","srashta","immanuel","beulah","prudhvi",np.nan],index=["Name "+str(i) for i in range(1,7)],name="Family")
print(s.hasnans)
```

```python
#empty
print(s.empty)
s = pd.Series()
print(s.empty)
```

```python
#is_unique
s = pd.Series(["yaswanth","srashta","immanuel","beulah","prudhvi",np.nan],index=["Name "+str(i) for i in range(1,7)],name="Family")
print(s.is_unique)
s = pd.Series(["yaswanth","srashta","immanuel","beulah","prudhvi",np.nan,"yaswanth"],name="Family")
print(s.is_unique)
```

```python
#nbytes
print(s.nbytes)
```

```python
#accesing values
print(s[0])
print(s[["Name 2","Name 3","Name 4"]])
print(s[:-1])
print(s[:])
```

```python
print(s.iloc[0],s.iloc[1])
print(s.iloc[:-1])
print(s.iloc[:])
```

```python
s = pd.Series(["yaswanth","srashta","immanuel","beulah","prudhvi",np.nan],index=["Name "+str(i) for i in range(1,7)],name="Family")
print(s["Name 2"])
print(s[["Name 1","Name 2","Name 4"]])
```

```python
#slicing on programmer defined indexes
print(s["Name 1":"Name 4"])
```

```python
print(s[::-1]) #normal slicing
```

***Setting the values of series object***

```python
s["Name 6"] = "Bhargav"
print(s)
```

```python
s = s.reindex(["Name 1","Name 2","Name 3","Name 4","Name 5","Name 6","Name 7","Name 8","Name 9"])
print(s)
s[["Name 7","Name 8","Name 9"]] = np.nan,np.nan,np.nan
s["Name 7":"Name 9"] = "Harshitha","Charan", "Ahad"
print(s)
```
